We want to study now a confidence interval for $\theta$ in the exponential model, for a sample of any size $n$, possibly small (for example $n = 10$). For $0 < p < 1$, $0 < \alpha < 1$, and $z_q^{(\nu)}$ the quantile of order $q$ ($0 < q < 1$) of the $\chi^2_\nu$ distribution, set:

$$I_n(\alpha, p) = \left[ \frac{z_{p\alpha}^{(2n)}}{2n\bar{X}_n},\; \frac{z_{1-(1-p)\alpha}^{(2n)}}{2n\bar{X}_n} \right].$$

The pivot is $2n\theta\bar{X}_n \sim \chi^2_{2n}$, hence
$$P\big(\theta \in I_n(\alpha,p)\big) = P\big(z_{p\alpha}^{(2n)} \le 2n\theta\bar{X}_n \le z_{1-(1-p)\alpha}^{(2n)}\big) = 1 - (1-p)\alpha - p\alpha = 1-\alpha.$$

We say that $I_n(\alpha,p)$ is a confidence interval of level $1-\alpha$ for the unknown parameter $\theta$ and any $n$. The parameter $p$ distributes the risk $\alpha$ between the left tail (risk $p\alpha$) and the right tail (risk $(1-p)\alpha$); $p = 1/2$ gives symmetric risks.

*Remark: the statement writes the lower bound with $z_\alpha^{(2n)}$; the lower bound consistent with the level $1-\alpha$ is $z_{p\alpha}^{(2n)}$, as used here and in the code.*
